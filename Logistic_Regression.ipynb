## 4. Logistic Regression Model

The model first calculates

$$
z = Xw+b
$$

and then

$$
\hat y = \sigma(z)
$$

where

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

If `X.shape = (m,n)` and `w.shape = (n,)`, then `X @ w` has shape `(m,)`.

## 5. Step 1 — Implement Sigmoid

### Task
Complete `sigmoid(z)`.

Use `np.exp()`. Your function should work for both a scalar and a NumPy array.

````python
def sigmoid(z):
    return 1/(1+np.exp(-z))
````

````python
test_values = np.array([-10, -2, 0, 2, 10])
print(sigmoid(test_values))

# Remember: sigmoid(0) should be 0.5
````

## 6. Step 2 — Compute the Linear Score

Implement

`z = X @ w + b`

### Hint
Use the `@` operator for matrix-vector multiplication.

````python
def compute_z(X, w, b):
    # YOUR CODE HERE
    return X@w + b
````

````python
w = np.zeros(X.shape[1])
b = 0.0

z = compute_z(X, w, b)

print("z:", z)
print("z shape:", z.shape)
````

## 7. Step 3 — Compute Predicted Probabilities

The predicted probability is

$$
\hat y = \sigma(Xw+b)
$$

### Task
Implement `predict_probability()` using the functions you have already written.

````python
def predict_probability(X, w, b):
    # YOUR CODE HERE
    temp0 = compute_z(X, w, b)
    y_hat = sigmoid(temp0)
    return y_hat
````

````python
probabilities = predict_probability(X, w, b)
print(probabilities)
````

## 8. Step 4 — Implement the Cost Function

For binary Logistic Regression,

$$
J =
-\frac{1}{m}\sum_{i=1}^{m}
\left[
y^{(i)}\log(\hat y^{(i)})
+
(1-y^{(i)})\log(1-\hat y^{(i)})
\right]
$$

Use `np.clip()` to avoid taking `log(0)`.

### Hint
```python
epsilon = 1e-15
probabilities = np.clip(probabilities, epsilon, 1-epsilon)
```

````python
def compute_cost(X, y, w, b):
    m = X.shape[0]

    # YOUR CODE HERE
    epsilon = 1e-15
    probabilities = predict_probability(X, w, b)
    probabilities = np.clip(probabilities, epsilon, 1-epsilon)
    cost = (-1/m) * (np.sum(y*np.log(probabilities) + (1-y)*np.log(1-probabilities)))

    return cost
````

````python
initial_cost = compute_cost(X, y, w, b)
print("Initial cost:", initial_cost)

# With w=0 and b=0, every probability is 0.5.
````

## 9. Step 5 — Compute the Gradients

The gradients are

$$
dw = \frac{1}{m}X^T(\hat y-y)
$$

$$
db = \frac{1}{m}\sum(\hat y-y)
$$

### Shape check

If

```text
X.shape = (m,n)
w.shape = (n,)
y.shape = (m,)
```

then

```text
X.T.shape = (n,m)
y_hat-y   = (m,)
dw        = (n,)
```

Do not use a loop over training examples. Rather, use broadcasting.

````python
def compute_gradients(X, y, w, b):
    m = X.shape[0]

    probabilities = predict_probability(X, w, b)
    dw = (1/m) * (X.T @ (probabilities - y))   # (n,m) @ (m,) -> (n,)
    db = (1/m) * np.sum(probabilities - y)

    return dw, db
````

````python
dw, db = compute_gradients(X, y, w, b)

print("dw:", dw)
print("dw shape:", dw.shape)
print("db:", db)
````

## 10. Step 6 — Implement One Gradient-Descent Update

Use

$$
w := w - \alpha dw
$$

$$
b := b - \alpha db
$$

### Task
Compute the gradients, update `w` and `b`, and return them.

````python
def gradient_descent_step(X, y, w, b, learning_rate):
    # YOUR CODE HERE
    dw, db = compute_gradients(X, y, w, b)
    w = w - learning_rate * dw
    b = b - learning_rate * db

    return w, b
````

## 11. Step 7 — Implement Complete Gradient Descent

For each iteration:
1. Compute probabilities.
2. Compute cost.
3. Compute gradients.
4. Update parameters.
5. Store the cost.

````python
def train_logistic_regression(X, y, learning_rate=0.1, num_iterations=1000):

    w = np.zeros(X.shape[1])
    b = 0.0

    cost_history = []

    for i in range(num_iterations):

        # YOUR CODE HERE
        cost = compute_cost(X, y, w, b)                       # cost at current parameters
        w, b = gradient_descent_step(X, y, w, b, learning_rate)  # gradients + update

        cost_history.append(cost)

    return w, b, cost_history
````

## 12. Train the Model

Use learning rate `0.1` and `1000` iterations.

````python
learning_rate = 0.1
num_iterations = 1000

w, b, cost_history = train_logistic_regression(
    X, y, learning_rate, num_iterations
)

print("Final weights:", w)
print("Final bias:", b)
print("Final cost:", cost_history[-1])
````

## 13. Plot the Cost

Plot iteration number against cost.

````python
plt.figure(figsize=(7,5))

# YOUR CODE HERE
plt.plot(range(len(cost_history)), cost_history)

plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost During Gradient Descent")
plt.show()
````

## 14. Step 8 — Convert Probabilities into Class Predictions

Use the rule

$$
\hat y =
\begin{cases}
1 & \text{if probability} \geq 0.5\\
0 & \text{otherwise}
\end{cases}
$$

### Hint
```python
(probabilities >= threshold).astype(int)
```

````python
def predict(X, w, b, threshold=0.5):
    # YOUR CODE HERE
    probabilities = predict_probability(X, w, b)
    return (probabilities >= threshold).astype(int)
````

````python
y_pred = predict(X, w, b)

print("Predictions:", y_pred)
print("Actual:     ", y)
````

## 15. Step 9 — Calculate Accuracy

$$
Accuracy =
\frac{\text{Number of correct predictions}}
{\text{Total number of predictions}}
$$

### Task
Implement `accuracy()` using NumPy.

````python
def accuracy(y_true, y_pred):
    # YOUR CODE HERE
    return np.mean(y_true == y_pred)
````

````python
print("Accuracy:", accuracy(y, y_pred))
````

## 16. Step 10 — Visualize the Decision Boundary

For two features, the decision boundary is

$$
w_1x_1+w_2x_2+b=0
$$

Therefore,

$$
x_2=-\frac{w_1x_1+b}{w_2}
$$

### Task
Plot the two classes and the learned decision boundary.

````python
plt.figure(figsize=(7,5))

# YOUR CODE HERE
plt.scatter(X[y==0,0], X[y==0,1], c="tab:blue", marker="o", label="Class 0")
plt.scatter(X[y==1,0], X[y==1,1], c="tab:red", marker="x", label="Class 1")

x1_vals = np.linspace(X[:,0].min()-0.5, X[:,0].max()+0.5, 100)
x2_vals = -(w[0]*x1_vals + b) / w[1]
plt.plot(x1_vals, x2_vals, "k--", label="Decision boundary")
plt.ylim(X[:,1].min()-0.5, X[:,1].max()+0.5)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Logistic Regression Decision Boundary")
plt.legend()
plt.show()
````

Train the model using:

### Question
Which learning rate appears most appropriate for this dataset, and why?

````python
learning_rates = [0.001, 0.1, 1.0]

# YOUR CODE HERE
for lr in learning_rates:
    w_lr, b_lr, hist = train_logistic_regression(X, y, lr, 1000)
    acc = accuracy(y, predict(X, w_lr, b_lr))
    print(f"lr={lr:<6} final cost={hist[-1]:.4f}  accuracy={acc:.4f}")
````

**Answer (Learning rate):**

|---|---|---|
| 0.001 | 0.6199 | 0.50 |
| 0.1 | 0.2089 | 0.9167 |
| 1.0 | 0.1001 | 1.00 |

- `0.001` is too small: after 1000 iterations the cost has barely moved from the starting value (0.693), so the model is under-trained and still predicts poorly.
- `1.0` ends with the lowest cost and perfect accuracy, but the cost is **not monotonically decreasing**. It overshoots early (cost rises above its starting value) before settling. Such large steps are risky and could diverge on other data.
- `0.1` decreases the cost smoothly and steadily and gives a good fit (cost 0.209, accuracy 0.917) without any instability. It is the most appropriate and safest choice for this dataset: it converges reliably. (If we only cared about final training fit, `1.0` wins here, and `0.1` would also reach it with more iterations.)

# Test your understanding — Classification Threshold

Use thresholds:

- `0.3`
- `0.5`
- `0.7`

Compare the predictions and accuracy.

### Question
Why can changing the threshold change the classification result without changing the learned weights?

````python
thresholds = [0.3, 0.5, 0.7]

# YOUR CODE HERE
for t in thresholds:
    preds = predict(X, w, b, threshold=t)
    print(f"threshold={t}: predictions={preds}  accuracy={accuracy(y, preds):.4f}")
````

**Answer (Threshold):** The weights fix the predicted *probabilities* only. The threshold is a separate decision rule applied afterwards to turn probabilities into classes. A lower threshold (0.3) labels more points as class 1, a higher one (0.7) labels fewer, so borderline points near the boundary flip class even though `w` and `b` are unchanged. Geometrically, the threshold shifts the decision boundary parallel to itself (boundary at `w·x + b = log(t/(1-t))`).
